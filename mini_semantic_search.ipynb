```python
# 1. Install the required libraries
!pip install -q pandas chromadb sentence-transformers

# 2. Define data, create dataframe, and save to CSV
import pandas as pd

data = {
    "question": [
        "How do I reset my password?",
        "What are your working hours?",
        "Where is the campus located?",
        "How do I contact technical support?"
    ],
    "answer": [
        "Go to settings and click 'Reset Password' to receive an email link.",
        "We are open Monday through Friday from 9 AM to 5 PM.",
        "The main campus is located in Addis Ababa.",
        "You can reach support via email at support@example.com."
    ]
}

df = pd.DataFrame(data)
df.to_csv("faqs.csv", index=False)

print("Libraries installed and 'faqs.csv' created successfully!")
display(df)
```

```python
import chromadb

# 1. Initialize the Chroma client
client = chromadb.Client()

# 2. Create a collection (Chroma automatically handles default text embeddings)
collection = client.create_collection(name="faq_collection")

# 3. Prepare our data from the pandas dataframe
documents = df["answer"].tolist()
metadatas = [{"question": q} for q in df["question"]]
ids = [str(i) for i in range(len(df))]

# 4. Add data to Chroma (this embeds the text vectors behind the scenes)
collection.add(
    documents=documents,
    metadatas=metadatas,
    ids=ids
)

print("Data successfully embedded and stored in ChromaDB!\n")

# 5. Query the database with a semantic search phrase
# Notice how we use "I forgot my login credentials" instead of the exact words "reset my password"
query_text = "I forgot my login credentials"

results = collection.query(
    query_texts=[query_text],
    n_results=3  # Get top 3 matches
)

# 6. Print the top 3 matches nicely
print(f"Query: '{query_text}'\n" + "="*40)
for i in range(len(results['documents'][0])):
    print(f"Match {i+1}:")
    print(f"  Original Question: {results['metadatas'][0][i]['question']}")
    print(f"  Answer: {results['documents'][0][i]}")
    print("-" * 40)
```
